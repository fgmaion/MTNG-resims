```python
GAMA = np.array([
    [6.875, -0.691, 0.176],
    [7.125, -1.084, 0.125],
    [7.375, -1.011, 0.071],
    [7.625, -1.349, 0.092],
    [7.875, -1.287, 0.079],
    [8.125, -1.544, 0.071],
    [8.375, -1.669, 0.045],
    [8.625, -1.688, 0.032],
    [9.125, -1.886, 0.020],
    [9.375, -2.055, 0.014],
    [9.625, -2.142, 0.010],
    [9.875, -2.219, 0.009],
    [10.125, -2.274, 0.009],
    [10.375, -2.292, 0.009],
    [10.625, -2.361, 0.010],
    [10.875, -2.561, 0.013],
    [11.125, -2.922, 0.019],
    [11.375, -3.414, 0.032],
    [11.625, -4.704, 0.138]
])

GAMA_corr = np.zeros((GAMA.shape[0], 4))
GAMA_corr[:,0] = np.log10( 10**GAMA[:,0] / 0.7**2 )
GAMA_corr[:,1] = np.log10( 10**GAMA[:,1] * 0.7**3 )

# Transforming the errors is a bit more involved
xn = 10**GAMA[:,1] * ( 1 - 10**(-GAMA[:,2]) )
xp = 10**GAMA[:,1] * ( 10**(GAMA[:,2]) - 1 )

GAMA_corr[:,2] = GAMA[:,1] - np.log10( 10**GAMA[:,1] - xn  )
GAMA_corr[:,3] = np.log10( 10**GAMA[:,1] + xp  ) - GAMA[:,1]
```
