```python
def get_normalized_image(image, vmin=None, vmax=None):
    if vmin is None:
        vmin = np.min(image)
    if vmax is None:
        vmax = np.max(image)

    image = np.clip(image, vmin, vmax)
    image = (image - vmin) / (vmax - vmin)

    return image
```
